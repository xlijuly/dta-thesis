LLM Scores (PC, BWS) -> long format table

```python
import pandas as pd
```

```python
# columns: tweet_id, model, method, score

dimensions = ['valence', 'arousal', 'dominance']
models = ['gpt', 'mini', 'nano']
methods = ['pc', 'bws']

rows = []

for model in models:
    for method in methods:
        for dimension in dimensions:
            file_path = f"{model}/{method}/{dimension}.csv"
            df = pd.read_csv(file_path, dtype={"Word": str})
            row = pd.DataFrame({
                "tweet_id": df["Word"].astype(str),
                "dimension": dimension,
                "method": method,
                "model": model,
                "score": df["RW"]
            })
            rows.append(row)

df = pd.concat(rows, ignore_index=True)
df.to_csv("llm_pc_bws_scores.csv", index=False)
```
