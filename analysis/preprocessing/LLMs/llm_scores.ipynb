Creating a long format table with all LLM-generated scores:
- Read previously generated long format tables with RS scores and PC&BWS scores;
- normalize RS scores to [0,1];
- merge the two tables;
- Data inspection

```python
import pandas as pd
```

Read and inspect the data:

```python
rs = pd.read_csv('rs/llm_rs_scores.csv')
# normalize the score to [0, 1]
rs['score'] = rs['score'] / 100
rs.head()
```

```python
pc_bws = pd.read_csv('llm_pc_bws_scores.csv')
# Concatenate the dataframes
df = pd.concat([rs, pc_bws], ignore_index=True)
```

```python
# count the unique values in each column except 'score'
for col in df.columns:
    if col != 'score':
        print(f"Unique values in column '{col}': {df[col].nunique()}")

# more necessary inspections can be added here
```

```python
# save the combined dataframe for later use
df.to_csv('llm_scores.csv', index=False)
```
