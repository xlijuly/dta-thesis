Summarizing LLM annotations with RS in one dataframe
- RS(0-100) or RS(0-10)

```python
import os
import json
import pandas as pd
```

```python
dimensions = ['valence', 'arousal', 'dominance']
models = ['gpt', 'mini', 'nano']
json_files = [f"rs10/{model}_rs10.json" for model in models]

mapping = pd.read_csv("rs_itemid_tweetid.csv", dtype={"item_id": str, "tweet_id": str})
rows = []

for model, json_file in zip(models, json_files):
    with open(json_file, 'r') as f:
        json_data = json.load(f)

    for entry in json_data:
        item_id = entry["item_id"]
        tweet_id = f"['{mapping.loc[mapping['item_id'] == item_id, 'tweet_id'].values[0]}']"

        for dimension in dimensions:
            rows.append({
                "tweet_id": tweet_id,
                "dimension": dimension,
                "method": "rs10",
                "model": model,
                "score": entry[dimension]
            })

df = pd.DataFrame(rows)
df.head()
```

```python
df.to_csv("llm_rs10_scores.csv", index=False)
```
