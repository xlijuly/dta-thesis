Notebook for turning PC/BWS annotations from LLMs (raw output in JSON format) into dataframes ready for converting into real-value scores (with the scripts from Hollis).
- added back the square brackets and single quotations around the tweet ids here: to make sure that the Hollis script parse them as strings (most important), and to make it easier later to match the ids in VAD_scores.

```python
import pandas as pd
import json
```

For PC:

```python
models = ["GPT-5.4_29Apr", "GPT-5.4-mini_29Apr", "GPT-5.4-nano_29Apr"]
dimensions = ["valence", "arousal", "dominance"]
pc_map = pd.read_csv("pc_itemid_tweetids.csv", dtype=str)

for model in models: 
    pc_outputs = pd.read_json(f"{model}/pc_outputs.json", dtype=str)
    for dimension in dimensions:
        hollis_input = []

        for idx, row in pc_map.iterrows():
            item_id = row["item_id"]
            tweet_id_1 = row["tweet_id_1"]
            tweet_id_2 = row["tweet_id_2"]
            
            output_row = pc_outputs[pc_outputs["item_id"] == item_id].iloc[0]
            best_id = output_row[f"{dimension}_best"]
            if best_id == tweet_id_1:
                worst_id = tweet_id_2  
            else:
                worst_id = tweet_id_1

            hollis_input.append({
                "best": f"['{best_id}']",
                "worst": f"['{worst_id}']",
                "option1": f"['{tweet_id_1}']",
                "option2": f"['{tweet_id_2}']"
            })

        hollis_df = pd.DataFrame(hollis_input,columns=["best", "worst", "option1", "option2"])
        hollis_df.to_csv(f"pc_hollis_input/{model}_{dimension}_pc_hollis_input.csv", index=False)

    
```

For BWS:

```python
bws_map = pd.read_csv("bws_itemid_tweetids.csv", dtype=str)

for model in models: 
    bws_outputs = pd.read_json(f"{model}/bws_outputs.json", dtype=str)
    for dimension in dimensions:
        hollis_input = []

        for idx, row in bws_map.iterrows():
            item_id = row["item_id"]
            tweet_id_1 = row["tweet_id_1"]
            tweet_id_2 = row["tweet_id_2"]
            tweet_id_3 = row["tweet_id_3"]
            tweet_id_4 = row["tweet_id_4"]
            
            output_row = bws_outputs[bws_outputs["item_id"] == item_id].iloc[0]
            best_id = output_row[f"{dimension}_best"]
            worst_id = output_row[f"{dimension}_worst"]

            hollis_input.append({
                "best": f"['{best_id}']",
                "worst": f"['{worst_id}']",
                "option1": f"['{tweet_id_1}']",
                "option2": f"['{tweet_id_2}']",
                "option3": f"['{tweet_id_3}']",
                "option4": f"['{tweet_id_4}']"
            })

        hollis_df = pd.DataFrame(hollis_input,columns=["best", "worst", "option1", "option2", "option3", "option4"])
        hollis_df.to_csv(f"bws_hollis_input/{model}_{dimension}_bws_hollis_input.csv", index=False)

    
```
